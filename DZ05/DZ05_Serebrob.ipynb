```python
import numpy as np
from matplotlib import pyplot as plt
from pyscf import gto, scf, mp
```

```python
def calc_N2_energies(basis, verbose=0):
    # N2 mol
    mol_N2 = gto.M(atom=f'N 0 0 0; N 0 0 1.095', basis=basis, verbose=verbose)
    mf_N2  = scf.RHF(mol_N2).run()
    E_N2_hf  = mf_N2.e_tot
    E_N2_mp2 = mp.MP2(mf_N2).run().e_tot

  # N atom -> Elec = 3
    mol_N = gto.M(atom='N 0 0 0', basis=basis, spin=3, verbose=verbose)
    mf_N  = scf.UHF(mol_N).run()
    E_N_hf  = mf_N.e_tot
    E_N_mp2 = mp.MP2(mf_N).run().e_tot

    # N atom + Dummy for BSSE
    mol_N_BSSE = gto.M(atom=f'N 0 0 0; ghost:N 0 0 1.095', basis=basis, spin=3, verbose=verbose)
    mf_N_BSSE  = scf.UHF(mol_N_BSSE).run()
    E_N_BSSE_hf  = mf_N_BSSE.e_tot
    E_N_BSSE_mp2 = mp.MP2(mf_N_BSSE).run().e_tot

    return dict(E_N2_hf=E_N2_hf, E_N2_mp2=E_N2_mp2,
                E_N_hf=E_N_hf,   E_N_mp2=E_N_mp2,
                E_N_BSSE_hf=E_N_BSSE_hf, E_N_BSSE_mp2=E_N_BSSE_mp2)
```

Экспериментально энергия связи в диазоте равна: 945.33(59) КДж/моль ==   0.36006(22) Eh

```python
Basis_sets = ['sto-6g', 'def2-svp']

Exp = 0.3600622
for basis in Basis_sets:
    print(f'Энергии связи для {basis} базиса: \n')
    e = calc_N2_energies(basis)
    E_hf = 2*e['E_N_hf'] - e['E_N2_hf']
    E_mp2 =  2*e['E_N_mp2'] - e['E_N2_mp2']
    Diff_hf = Exp - E_hf
    Diff_mp2 = Exp - E_mp2
    E_hf_BSSE = 2*e['E_N_BSSE_hf'] - e['E_N2_hf']
    E_mp2_BSSE =  2*e['E_N_BSSE_mp2'] - e['E_N2_mp2']
    Diff_BSSE_hf = Exp - E_hf_BSSE
    Diff_BSSE_mp2 = Exp - E_mp2_BSSE
    print(f'Энергия связи N-N в HF : {E_hf}, в MP2 : {E_mp2} \n')
    print(f'Энергии с поправкой на BSSE\n HF : {E_hf_BSSE}, MP2 : {E_mp2_BSSE} \n ')
    print(f'Разница между расчетом и экспериментом: \nHF : {Diff_hf}\nMP2 : {Diff_mp2}\nHF + BSSE: {Diff_BSSE_hf}\nMP2 + BSSE: {Diff_BSSE_mp2}\n ')
```
